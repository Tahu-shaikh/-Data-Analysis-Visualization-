```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
customer_sales = df.groupby('Customer_Type')['Total'].sum()

customer_sales.plot(
    kind='bar',
    color=['blue','orange']
)

plt.title("Revenue by Customer Type")
plt.xlabel("Customer Type")
plt.ylabel("Revenue")

plt.savefig(
    "visualizations/customer_type_spending.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
gender_sales = df.groupby('Gender')['Total'].sum()

gender_sales.plot(
    kind='pie',
    autopct='%1.1f%%'
)

plt.title("Revenue by Gender")
plt.ylabel("")

plt.savefig(
    "visualizations/gender_spending.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
import seaborn as sns

plt.figure(figsize=(8,5))

sns.histplot(
    df['Rating'],
    bins=10,
    kde=True
)

plt.title("Customer Rating Distribution")
plt.xlabel("Rating")
plt.ylabel("Frequency")

plt.savefig(
    "visualizations/customer_rating_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
payment_method = df['Payment'].value_counts()

payment_method.plot(
    kind='bar',
    color='green'
)

plt.title("Preferred Payment Methods")
plt.xlabel("Payment Method")
plt.ylabel("Count")

plt.savefig(
    "visualizations/payment_method_by_customer.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
corr = df[['Total','Quantity','Rating','Unit_Price']].corr()

plt.figure(figsize=(8,6))

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm'
)

plt.title("Customer Behavior Correlation Heatmap")

plt.savefig(
    "visualizations/customer_behavior_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
