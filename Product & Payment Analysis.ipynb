```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("supermarket_sales_cleaned.csv")
```

```python
product_revenue = (
    df.groupby('Product_Line')['Total']
    .sum()
    .sort_values(ascending=False)
)

print(product_revenue)
```

```python
product_revenue.plot(
    kind='bar',
    color='skyblue'
)

plt.title("Revenue by Product Line")
plt.xlabel("Product Line")
plt.ylabel("Revenue")

plt.tight_layout()

plt.savefig(
    "visualizations/product_revenue.png",
    dpi=300
)

plt.show()
```

```python
quantity_sold = (
    df.groupby('Product_Line')['Quantity']
    .sum()
    .sort_values(ascending=False)
)

print(quantity_sold)
```

```python
quantity_sold.plot(
    kind='bar',
    color='green'
)

plt.title("Quantity Sold by Product Line")
plt.xlabel("Product Line")
plt.ylabel("Quantity")

plt.tight_layout()

plt.savefig(
    "visualizations/quantity_sold.png",
    dpi=300
)

plt.show()
```

```python
payment_revenue = (
    df.groupby('Payment')['Total']
    .sum()
)

print(payment_revenue)
```

```python
payment_revenue.plot(
    kind='pie',
    autopct='%1.1f%%'
)

plt.title("Revenue by Payment Method")
plt.ylabel("")

plt.tight_layout()

plt.savefig(
    "visualizations/payment_revenue.png",
    dpi=300
)

plt.show()
```

```python
product_rating = (
    df.groupby('Product_Line')['Rating']
    .mean()
    .sort_values(ascending=False)
)

print(product_rating)
```

```python
product_rating.plot(
    kind='bar',
    color='orange'
)

plt.title("Average Rating by Product Line")
plt.xlabel("Product Line")
plt.ylabel("Average Rating")

plt.tight_layout()

plt.savefig(
    "visualizations/product_rating.png",
    dpi=300
)

plt.show()
```

```python
monthly_sales = pd.pivot_table(
    df,
    values='Total',
    index='Month',
    columns='Product_Line',
    aggfunc='sum'
)

monthly_sales
```

```python
monthly_sales.plot(
    figsize=(12,6)
)

plt.title("Monthly Product Sales Trends")
plt.xlabel("Month")
plt.ylabel("Revenue")

plt.tight_layout()

plt.savefig(
    "visualizations/monthly_product_sales.png",
    dpi=300
)

plt.show()
```

```python
product_rating = (
    df.groupby('Product_Line')['Rating']
    .mean()
    .sort_values(ascending=False)
)

print(product_rating)
```

```python
corr = df[['Unit_Price','Quantity','Tax','Total','Rating']].corr()

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm'
)

plt.show()
```
